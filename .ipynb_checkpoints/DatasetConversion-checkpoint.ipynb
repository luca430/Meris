# Null model data

```julia
S = 50 # n species
T = 100_000 # n samples
N = 20_000 # n reads

Random.seed!(1234)

# invent unique IDs
species_ids = ["species_$(i)" for i in 1:S]
sample_ids  = ["s$j"  for j in 1:T]
# p = rand(Dirichlet(rand(Gamma(10,2), S)))
# p = rand(Dirichlet(rand(Float64, S)))
p = rand(Dirichlet([S for _ in 1:S]))

N_vec  = floor.(rand(LogNormal(log(N), 1), T))
# N_vec  = floor.(rand(Gamma(10, 2*N), T))
samples = [[rand(Poisson(N_vec[t] * p[s])) for s in 1:S] for t in 1:T]
count_mat = hcat(samples...)

df_wide = DataFrame(count_mat, sample_ids)
df_wide.species_id .= species_ids

NULL_MODEL = stack(df_wide, Not(:species_id); variable_name = :sample_id, value_name = :count)

# add the nreads column by lookup
idx = parse.(Int, replace.(NULL_MODEL.sample_id, r"^s" => ""))
NULL_MODEL.nreads = N_vec[idx]
NULL_MODEL.env .= "NULL"

# reorder columns if you like
select!(NULL_MODEL, [:env, :species_id, :sample_id, :count, :nreads])

@save "./Data/processed/null_model.jld2" NULL_MODEL
```

# Mobility data

```julia
fname = "./Data/mobility_raw/loc-gowalla_totalCheckins.txt.gz"
    
# Open a decompression stream over the gz file
gz = GzipDecompressorStream(open(fname, "r"))
colnames = [:user_id, :time_z, :lat, :long, :location_id]

# Read into a DataFrame
df_raw = CSV.File(
    gz;
    delim='\t',
    header=colnames,
    strict=false,    # allow rows with varying column counts
) |> DataFrame

close(gz)

dates = [row.time_z[1:10] for row in eachrow(df_raw)]
df_raw.time = dates
MOBILITY = select(df_raw, [:user_id, :location_id, :time])

MOBILITY = transform(
  groupby(MOBILITY, [:time, :location_id]),
  :user_id => length => :count
)

MOBILITY = transform(
  groupby(MOBILITY, [:time]),
  :count => sum => :nreads
)

rename!(MOBILITY, :location_id => :species_id, :time => :sample_id)
MOBILITY.env .= "CHECK-IN"
select!(MOBILITY, [:env, :species_id, :sample_id, :count, :nreads])

@save "./Data/processed/mobility.jld2" MOBILITY
```

# Books counts data

```julia
df = CSV.read("./Data/books_raw/gutenberg_data.csv", DataFrame)
select!(df, [:ID, :Bookshelf, :Text])

# define your mapping
category_map = Dict(
    # --- animals ---
    "Animal"                      => "ANIMALS",
    "Animals-Domestic"            => "ANIMALS",
    "Animals-Wild"                => "ANIMALS",
    "Animals-Wild-Trapping"       => "ANIMALS",
    "Animals-Wild-Birds"          => "ANIMALS",
    "Animals-Wild-Insects"        => "ANIMALS",
    "Animals-Wild-Mammals"        => "ANIMALS",
    "Animals-Wild-Reptiles"       => "ANIMALS",

    # --- geography ---
    "Canada"                      => "GEOGRAPHY",
    "Italy"                       => "GEOGRAPHY",
    "United"                      => "GEOGRAPHY",
    "Norway"                      => "GEOGRAPHY",
    "Bulgaria"                    => "GEOGRAPHY",
    "Argentina"                   => "GEOGRAPHY",
    "South"                       => "GEOGRAPHY",
    "France"                      => "GEOGRAPHY",
    "Greece"                      => "GEOGRAPHY",
    "Germany"                     => "GEOGRAPHY",
    "Australia"                   => "GEOGRAPHY",
    "Egypt"                       => "GEOGRAPHY",
    "India"                       => "GEOGRAPHY",
    "Africa"                      => "GEOGRAPHY",

    # --- holidays ---
    "Christmas"                   => "HOLIDAYS",
    "Travel"                      => "HOLIDAYS",

    # --- education ---
    "School"                      => "EDUCATION",
    "Harvard"                     => "EDUCATION",

    # --- classical
    "Classical"                   => "CLASSICAL",

    # --- other children ---
    "Children's"                  => "CHILDREN",
    "Child's"                     => "CHILDREN",
    
    # --- fiction ---
    "Crime"                       => "FICTION",
    "Mystery"                     => "FICTION",
    "Detective"                   => "FICTION",

    # --- Not defined ---
    "New"                         => "ND",
    "Women's"                     => "ND",
    "6"                           => "ND"
);

df.Bookshelf = get.(Ref(category_map), df.Bookshelf, "ND")
unique!(df, [:Bookshelf, :ID, :Text])

STOPWORDS = Set([
    "me","my","myself","we","our","ours","ourselves","you","your","yours","yourself","yourselves","he","him","his","himself",
    "she","her","hers","herself","it","its","itself","hey","them","their","theirs","themselves","what","which","who","whom",
    "this","that","these","those","am","is","are","was","were","be","been","being","have","has","had","having","do","does","did",
    "doing","an","the","and","but","if","or","because","as","until","while","of","at","by","for","with","about","against","between",
    "into","through","during","before","after","above","below","to","from","up","down","in","out","on","off","over","under","again",
    "further","then","once","here","there","when","where","why","how","all","any","both","each","few","more","most","other","some",
    "such","no","nor","not","only","own","same","so","than","too","very","can","will","just","don","should","now",
    "a","b","c","d","e","f","g","h","i","j","k","l","m","n","o","p","q","r","s","t","u","v", "w","x","y","z"
])

function word_counts(text; remove=true)
    # Normalize & split into all words (including stop-words)
    text = lowercase(text)
    # Split on anything that isn't A–Z or a–z (one or more in a row)
    raw_words = filter(!isempty, split(text, r"[^A-Za-z]+"))
    
    # Count everything
    cnt_map = countmap(raw_words)
    nreads = sum(values(cnt_map))
    
    if remove # Remove stop-words from the count map
        filter!(p -> !(p.first in STOPWORDS), cnt_map)
    end
    
    return cnt_map, nreads
end

# prepare empty DataFrame with the right column types
BOOKS2 = DataFrame(
    env        = String[],
    species_id = String[],
    sample_id  = Int[],
    count      = Int[],
    nreads     = Int[],
)

for row in eachrow(df)
    if !ismissing(row.Text)
        env, sid = row.Bookshelf, row.ID
        cnt_map, nreads = word_counts(row.Text, remove=false)
        for (word, freq) in cnt_map
            push!(BOOKS2, (env, word, sid, freq, nreads))
        end
    end
end

unique!(BOOKS2, [:env, :species_id, :sample_id])

@save "./Data/processed/books2.jld2" BOOKS2
```

# Cross-sectional microbiome data

```julia
function GetCrossSecData(path; min_samples=1, min_counts=1, min_nreads=1)
    
    raw_data = load(path)
    proj_time = raw_data["datatax"]
    
    # Fix environments with inverted nreads/count
    mask = proj_time.nreads .< proj_time.count
    for i in eachindex(mask)
        if mask[i]
            tmp = proj_time.nreads[i]
            proj_time.nreads[i] = proj_time.count[i]
            proj_time.count[i] = tmp
        end
    end
    
    # Filter by min_samples
    grouped = groupby(proj_time, [:project_id, :classification])
    filtered_data = combine(grouped) do sdf
        if length(unique(sdf.run_id)) >= min_samples
            return sdf
        else
            return DataFrame()
        end
    end
    
    # Filter by min_nreads and min_counts
    filtered_data = filtered_data[filtered_data.nreads .>= min_nreads, :]
    filtered_data = filtered_data[filtered_data.count .>= min_counts, :]
    
    # Separate by environment
    SEAWATER = filtered_data[filtered_data.classification .== " seawater", :]
    ORAL = filtered_data[filtered_data.classification .== "ORAL", :]
    GUT = filtered_data[filtered_data.classification .== "GUT", :]
    VAGINAL = filtered_data[filtered_data.classification .== "VAGINAL", :]
    SOIL = filtered_data[filtered_data.classification .== "Environmental Terrestrial Soil", :]
    ORALCAVITY = filtered_data[filtered_data.classification .== "oralcavity", :]
    FECES = filtered_data[filtered_data.classification .== "feces", :]
    SKIN = filtered_data[filtered_data.classification .== "skin", :]
    GLACIER = filtered_data[filtered_data.classification .== "Glacier", :]
    AQUA1 = filtered_data[filtered_data.classification .== "Environmental Aquatic Marine Hydrothermal vents", :]
    RIVER = filtered_data[filtered_data.classification .== "River", :]
    LAKE = filtered_data[filtered_data.classification .== "Lake", :]
    AQUA2 = filtered_data[filtered_data.classification .== "Environmental Aquatic Marine", :]
    SLUDGE = filtered_data[filtered_data.classification .== "activatedsludge", :]
    
    select!(SEAWATER, Not([:classification]))
    select!(ORAL, Not([:classification]))
    select!(GUT, Not([:classification]))
    select!(VAGINAL, Not([:classification]))
    select!(SOIL, Not([:classification]))
    select!(ORALCAVITY, Not([:classification]))
    select!(FECES, Not([:classification]))
    select!(SKIN, Not([:classification]))
    select!(GLACIER, Not([:classification]))
    select!(AQUA1, Not([:classification]))
    select!(RIVER, Not([:classification]))
    select!(LAKE, Not([:classification]))
    select!(AQUA2, Not([:classification]))
    select!(SLUDGE, Not([:classification]))
    
    sep_data = Dict("SEAWATER" => SEAWATER, "ORAL" => ORAL, "GUT" => GUT, "VAGINAL" => VAGINAL,
                         "SOIL" => SOIL, "ORALCAVITY" => ORALCAVITY, "FECES" => FECES, "SKIN" => SKIN,
                         "GLACIER" => GLACIER, "AQUA1" => AQUA1, "RIVER" => RIVER, "LAKE" => LAKE,
                         "AQUA2" => AQUA2, "SLUDGE" => SLUDGE)

    return sep_data
end

sep_data = GetCrossSecData("./Data/microbiome_raw/crosssecdata.RData"; min_samples=1, min_counts=1, min_nreads=1);

for key in keys(sep_data)
    df = copy(sep_data[key])
    select!(df, Not([:project_id, :sample_id]))
    rename!(df, :otu_id => :species_id, :run_id => :sample_id)
    df.env .= key
    select!(df, :env, :species_id, :sample_id, :count, :nreads)
    sep_data[key] = df
end

# Join multiple dfs together (if they have the same columns)
MICROBIOME = vcat(values(sep_data)...)

@save "./Data/processed/microbiome.jld2" MICROBIOME
```

# Genetic expression data

```julia
# Load file with metadata info to get nreads for each sample
fname = "./Data/genetic_raw/GTEx_Analysis_v10_Annotations_SampleAttributesDS.txt"
df_reads = CSV.read(fname, DataFrame; delim='\t')
df_reads = select(df_reads, :SAMPID, :SMRDTTL)
rename!(df_reads, :SAMPID => :sample_id, :SMRDTTL => :nreads)

# Loop over different datasets to make a unique DataFrame
fnames = ["brain_cortex", "heart_left_ventricle", "liver", "pancreas"]
df_names = ["BRAIN", "HEART", "LIVER", "PANCREAS"]
all_df = []

for i in 1:length(fnames)
    fname = "./Data/genetic_raw/gene_reads_v10_$(fnames[i]).gct.gz"
    
    # Open a decompression stream over the gz file
    gz = GzipDecompressorStream(open(fname, "r"))
    
    # Read into a DataFrame
    df_raw = CSV.File(
        gz;
        delim='\t',
        header=3,
        strict=false,    # allow rows with varying column counts
    ) |> DataFrame
    
    close(gz)
    
    select!(df_raw, Not([:Description]))
    df = stack(df_raw, Not(:Name); variable_name = :sample_id, value_name = :counts)
    filter!(:counts => x -> x .> 0, df)
    rename!(df, :Name => :species_id, :counts => :count)
    df.env .= df_names[i]
    select!(df, :env, :species_id, :sample_id, :count)
    df = leftjoin(df, df_reads, on = :sample_id)
    push!(all_df, df)
end
    
# Join multiple dfs together (if they have the same columns)
GENETIC = vcat(all_df...)

@save "./Data/processed/genetic.jld2" GENETIC
```

# Fossil occurances data

```julia
fnames = ["africa.csv", "asia.csv", "europe.csv", "north_america.csv", "oceania.csv", "south_america.csv"]
df_names = ["AFRICA", "ASIA", "EUROPE", "NORTH AMERICA", "OCEANIA", "SOUTH AMERICA"]
all_df = []

for i in 1:length(fnames)
    df = CSV.File("./Data/fossils_raw/$(fnames[i])"; header=17) |> DataFrame
    grouped = combine(groupby(df, [:collection_no, :accepted_name]), nrow => :count)
    rename!(grouped, :collection_no => :sample_id, :accepted_name => :species_id)
    grouped.env .= df_names[i]
    sample_sums = combine(groupby(grouped, :sample_id), :count => sum => :nreads)
    df = leftjoin(grouped, sample_sums, on = :sample_id)
    filter!(row -> !ismissing(row.sample_id) && !ismissing(row.species_id) && row.species_id != "", df)
    
    # Step 8: Reorder columns
    select!(df, :env, :species_id, :sample_id, :count, :nreads)
    
    # Step 9: Append to master collection
    push!(all_df, df)
end

# Join multiple dfs together (if they have the same columns)
FOSSILS = vcat(all_df...)

@save "./Data/processed/fossils.jld2" FOSSILS
```
