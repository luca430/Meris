```julia
include("./utils.jl")
using .Utils
```

```julia
T = Int(1e5)
N = Int(1e5)
p1, p2 = 1e-3, 2e-3
β = 5

Random.seed!(1234)

# N_vec = Int.(floor.(rand(Gamma(β, 2*N), T)))
N_vec = [N for _ in 1:T]

x1 = rand(Binomial(N,p1),T)
x2 = [rand(Binomial(N_vec[t], p2)) / N_vec[t] for t in 1:T];
```

```julia
y1 = log.(x1[x1 .> 0])
y2 = log.(x2[x2 .> 0])

z1 = (y1 .- mean(y1)) ./ std(y1)
z2 = (y2 .- mean(y2)) ./ std(y2)

data = [z1, z2]

hists = []
for d in data
    bmin, bmax = minimum(d), maximum(d)
    Δb = (bmax - bmin) / 30
    fh = FHist.Hist1D(d, binedges=bmin:Δb:bmax)
    ctrs = bincenters(fh)
    pdf  = bincounts(fh) ./ (integral(fh) * Δb)
    
    mask = pdf .> 0
    push!(hists, (ctrs[mask], pdf[mask]))
end;
```

```julia
fig = Figure(size=(1100,500))
ax  = Axis(fig[1, 1];
        yscale = log10, limits=((-5,4),(1e-8,1)),
        xlabel = "z", ylabel = "pdf", 
        title = "AFD")

for h in hists
    scatter!(ax, h[1], 10 .^ log.(h[2]))
end

zrange = minimum(vcat(data...)):1e-2:maximum(vcat(data...))
# lines!(ax, zrange, 10 .^ Utils.lrl(zrange, β), label="Lomax", color=:red)
lines!(ax, zrange, 10 .^ Utils.lre(zrange), label="Exponential", color=:blue, linewidth=2)
axislegend()
```

```julia
fig = Figure(size=(1100,500))
ax  = Axis(fig[1, 1];
        yscale=log10,
        xlabel = "z", ylabel = "pdf", 
        title = "AFD")

# Parameters
N = 100  # Large number of trials
p = 0.001  # Small probability
num_samples = 10000  # Number of samples to generate

# Simulate the binomial distribution
x_samples = Float64[]
for _ in 1:num_samples
    n = rand(Binomial(N, p))  # Sample from Binomial(N, p)
    push!(x_samples, n / N)   # Compute x = n / N
end

x_samples = log.(x_samples[x_samples .> 0])

# Plot the histogram of x
bmin, bmax = minimum(x_samples), maximum(x_samples)
Δb = (bmax - bmin) / 30
fh = FHist.Hist1D(x_samples, binedges=bmin:Δb:bmax)
ctrs = bincenters(fh)
pdf  = bincounts(fh) ./ (integral(fh) * Δb)

mask = pdf .> 0
ctrs = ctrs[mask]
pdf = pdf[mask]

scatter!(ax, ctrs, 10 .^ log.(pdf))

# Overlay the theoretical exponential distribution
λ = N * log((1 - p) / p)  # Rate parameter for the exponential distribution
x_vals = -9:1e-4:-1  # Values to plot the exponential distribution
y_vals = λ .* exp.(-λ .* exp.(x_vals)) .* exp.(x_vals)  # Exponential distribution PDF
lines!(ax, x_vals, 10 .^ log.(y_vals), label="Exponential Approximation", linewidth=2)
axislegend();
```
