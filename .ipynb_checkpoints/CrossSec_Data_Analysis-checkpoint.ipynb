### Correlations

```julia
vals = []
for i in 1:size(stats[4], 1)
    for j in 1:i
        push!(vals, stats[4][i,j])
    end
end


bmin = minimum(vals)
bmax = maximum(vals)
Δb = (bmax - bmin) / 40
fh = FHist.Hist1D(vals, binedges=bmin:Δb:bmax)

centers = bincenters(fh)
norm_counts = bincounts(fh) ./ (integral(fh) * Δb)

# Filter non-zero counts
valid = norm_counts .> 0.0
yy = log.(norm_counts[valid])
centers = centers[valid];
```

```julia
fig = Figure(figsize = (600, 400))
ax  = Axis(fig[1, 1]; 
           xlabel = "ρ", ylabel = "pdf", 
           title = "Pearson correlation")

scatter!(ax, centers, yy;
         color       = :tomato,    # marker color
         markersize  = 14,          # marker size in pixels
         strokewidth = 0.5,        # outline width
         strokecolor = :black)     # outline color

fig
```
