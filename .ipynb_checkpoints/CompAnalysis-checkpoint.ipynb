```julia
@load "./Data/AFD/microbiome_afd.jld2" afd_out
microbiome_afd = deepcopy(afd_out)

@load "./Data/AFD/genetic_afd.jld2" afd_out
genetic_afd = deepcopy(afd_out)

@load "./Data/AFD/books_afd.jld2" afd_out
books_afd = deepcopy(afd_out)

AFDs = [microbiome_afd, genetic_afd, books_afd];
```

```julia
fig = Figure(figsize=(900,500))
ax  = Axis(fig[1, 1];
        yscale=log10, limits=((-10.5,10),(1e-16,1e1)),
        xlabel = "log(z)", ylabel = "pdf", 
        title = "AFD")

shapes = (
    :circle,
    :rect,
    :diamond,
    :utriangle,
    :dtriangle,
    :xcross,
    :cross,
    :star5,
    :hexagon,
)

colors = (
    :orange,
    :seagreen1,
    :dodgerblue3
)

for (j,afd) in enumerate(AFDs)
    for (i,key) in enumerate(keys(afd))
        x, y = afd[key]
        sc = scatter!(ax, x, 10 .^ log.(y),
                    label=key,
                    color=colors[j],
                    marker=shapes[i],
                    markersize=12,
                    strokewidth = 0.8,
                    strokecolor = :black)
    end
end

#### ZOO OF DISTRIBUTIONS ###
function lrg(z, α)
    return α*sqrt(trigamma(α)) .* z .+ α*digamma(α) .- exp.(z .* sqrt(trigamma(α)) .+ digamma(α)) .+ 0.5*log(trigamma(α)) .- loggamma(α)
end

function lrb(z, α, β)
    m = digamma(α) - digamma(α + β)
    s = sqrt(trigamma(α) - trigamma(α + β))
    c = loggamma(α + β) - loggamma(α) - loggamma(β)
    return c .+ α .* (z .* s .+ m) .+ (β - 1) .* log.(1 .- exp.(z .* s .+ m)) .+ log(s)
end

function lrln(z, σ)
    return -z .^ 2 ./ 2 .+ log(sqrt(σ / 2 / π))
end

zrange = -10:0.01:10
gl = lines!(ax, zrange, 10 .^ lrg(zrange, 2.5), color=:red, linewidth=2.5, label="rescaled gamma")
bl = lines!(ax, zrange, 10 .^ lrb(zrange, 2.5, 5e3), color=:lime, linestyle=:dash, linewidth=2.5, label="rescaled beta")
lnl = lines!(ax, zrange, 10 .^ lrln(zrange, 2), color=:magenta, linestyle=:dashdot, linewidth=2.5, label="rescaled lognormal")
#############################

leg = Legend(fig, ax; orientation = :vertical)
fig[1, 2] = leg;
```
