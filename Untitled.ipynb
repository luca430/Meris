```julia
function pdf(x, N, p)
    r = p / (1 - p)
    C = N * log(r) / (r ^ N - 1)
    return C .* r .^ (N .* x)
end

function cdf(x, N, p)
    r = p / (1 - p)
    C = N * log(r) / (r ^ N - 1)
    return C ./ (N * log(r)) .* (r .^ (N .* x) .- 1)
end

function Exp01(N, p)
    u = rand(Float64)
    r = p / (1 - p)
    C = N * log(r) / (r ^ N - 1)
    return log(1 + u * N * log(r) / C) / (N * log(r))
end;
```

```julia
# Parameters
N = 1000  # Large number of trials
p = 0.01  # Small probability
num_samples = 10000  # Number of samples to generate

# 1. Exact Binomial Distribution (n ~ Binomial(N, p), then x = n / N)
binomial_x = Float64[]
for _ in 1:num_samples
    n = rand(Binomial(N, p))  # Sample from Binomial(N, p)
    push!(binomial_x, n / N)   # Compute x = n / N
end

# 2. Poisson Approximation (n ~ Poisson(N * p), then x = n / N)
poisson_x = Float64[]
λ = N * p  # Poisson rate parameter
for _ in 1:num_samples
    n = rand(Poisson(λ))  # Sample from Poisson(N * p)
    push!(poisson_x, n / N)   # Compute x = n / N
end

# 3. Exponential Approximation (x ~ Exp(N * log((1-p)/p)))
exp_x = [Exp01(N, p) for _ in 1:num_samples]

# Create histograms using FHist
function create_histogram(x_samples)
    bmin, bmax = minimum(x_samples), maximum(x_samples)
    Δb = (bmax - bmin) / 30
    fh = FHist.Hist1D(x_samples, binedges=bmin:Δb:bmax)
    ctrs = bincenters(fh)
    pdf = bincounts(fh) ./ (integral(fh) * Δb)

    # Masking to remove zero values
    mask = pdf .> 0
    ctrs = ctrs[mask]
    pdf = pdf[mask]

    return ctrs, pdf
end

# Calculate histograms
binomial_ctrs, binomial_pdf = create_histogram(binomial_x)
poisson_ctrs, poisson_pdf = create_histogram(poisson_x)
exp_ctrs, exp_pdf = create_histogram(exp_x)

# Plot using CairoMakie
fig = Figure(size=(800, 600))
ax = Axis(fig[1, 1];
        limits = ((0, 0.01), (1e-1, 5000)),
        yscale = log10,
        xlabel = "x", ylabel = "PDF", 
        title = "Comparison of Distributions")

# Plot all distributions as scatter plots
scatter!(ax, binomial_ctrs, binomial_pdf, label="Binomial x", color=:blue, markersize=10)
scatter!(ax, poisson_ctrs, poisson_pdf, label="Poisson x", color=:red, markersize=10)
scatter!(ax, exp_ctrs, exp_pdf, label="Exponential x", color=:green, markersize=10)

xrange=0:1e-4:0.05
lines!(ax, xrange, pdf(xrange, N, p))

# Add the legend
axislegend(ax)

# Display the figure
fig
```
