```julia
include("./utils.jl")
using .Utils
```

```julia
N = Int(2e4)
S = 50
T = 100

Random.seed!(1234)
Nv = Int.(floor.(rand(LogNormal(log(N), 1), T))) # length of books

### Compute Poisson rates ###
k = 1 # dispersion parameter
mv = rand(Dirichlet([1 for _ in 1:S])) # mean word-rates
λ_mat = [rand(Gamma(k, m/k), T) for m in mv] # real word-rates (each sample has its own real rate)
λ_mat = hcat(λ_mat...)

### Compute word counts and nreads ###
counts = Int.(zeros(T,S))
for i in 1:size(counts, 1)
    for j in 1:size(counts, 2)
        counts[i,j] = rand(Poisson(λ_mat[i,j] * Nv[i]))
    end
end

nreads = sum(counts, dims=2)

### Convert to dataframe ###
NULL_MODEL = DataFrame(
    env = Float64[],
    species_id = Int[],
    sample_id = Int[],
    count = Int[],
    nreads = Int[]
)

for i in 1:size(counts, 1)
    for j in 1:size(counts, 2)
        push!(NULL_MODEL, ("POISSON-GAMMA", j, i, counts[i,j], nreads[i]), promote=true)
    end
end

NULL_MODEL = NULL_MODEL[NULL_MODEL.count .!= 0, :];
```

```julia
### Pass dataframe to pipeline ###
Utils.df_filter!(NULL_MODEL; min_samples=5e1, min_nreads=1e4)

cutoffs_null = Dict("POISSON-GAMMA" => exp(-20))

println("--- AFD ---")
out = Utils.compute_AFD(NULL_MODEL; occ=1, bins=30, plot=false, verbose=true, save=true, filename="./Data/AFD/null_afd")
@load "./Data/AFD/null_afd.jld2" afd_out

println("--- TL ---")
out = Utils.compute_TL(NULL_MODEL; occ=0.9, bins=30, plot=false, verbose=true, save=true, filename="./Data/TL/null_tl", interpretation=2)
@load "./Data/TL/null_tl.jld2" taylor_out

println("--- MAD c ---")
out = Utils.compute_MAD(NULL_MODEL; c=cutoffs_null, bins=30, plot=false, verbose=true, save=true, filename="./Data/MAD/null_mad_c")
@load "./Data/MAD/null_mad_c.jld2" mad_out
mad_c_out = deepcopy(mad_out)

println("--- MAD ---")
out = Utils.compute_MAD(NULL_MODEL; c=nothing, bins=30, plot=false, verbose=true, save=true, filename="./Data/MAD/null_mad")
@load "./Data/MAD/null_mad.jld2" mad_out;
```
